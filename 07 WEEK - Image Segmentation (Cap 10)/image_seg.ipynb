Verificación y partición en regiones homogéneas

```python
import numpy as np

def evaluate_region(region, max_std):
    if len(region)== 0:
        return False, 0.0, 0.0
    mean_val = np.mean(region)
    std_val = np.std(region)
    es_homogenea = (std_val<=max_std)
    return es_homogenea, mean_val, std_val
```

```python
img = np.zeros((10,10), dtype=np.uint8)
img[1:5, 1:5] = 2 # Región R1
img[6:9, 6:9] = 8

R1 = img[1:5,1:5]

es_homogenea, media, std = evaluate_region(R1, 1)
print(f"Subregión R1 -> std: {std:.2f} | ¿Homogénea?: {es_homogenea}")

es_homogenea, media, std = evaluate_region(img, 0.5)
print(f"Subregión img -> std: {std:.2f} | ¿Homogénea?: {es_homogenea}")
```

```python
def vecinos(arreglo, fila, col):
    N = len(arreglo)
    M= len(arreglo[0]) if N > 0 else 0
    # fila, col = 1, 1
    vecindad =[]
    for f in range(max(0,fila-1),min(N-1,fila+1)+1):
        for c in range(max(0,col-1),min(M-1,col+1)+1):
            if f == fila and c == col:
                # arreglo[fila,col] = 0
                continue
            vecindad.append(float(arreglo[f][c]))
    return vecindad
```

```python
import numpy as np

def obtener_vecinos_etiquetados(mapa_bloques, i, j):
    """
    Extrae la vecindad 3x3 del píxel (i, j) y devuelve
    únicamente aquellos vecinos que YA tienen un número de bloque (> 0).
    """
    filas, cols = mapa_bloques.shape
    
    # Recorte 3x3 seguro dentro de los límites
    ventana = mapa_bloques[max(0, i - 1) : min(filas, i + 2),
                           max(0, j - 1) : min(cols,  j + 2)]
    
    # Extraemos solo los valores mayores a 0 (bloques asignados)
    vecinos_con_bloque = ventana[ventana > 0]
    
    return vecinos_con_bloque

def segmentar_bloques(imagen_binaria):
    """
    Recorre la imagen:
      - Si encuentra un 1 sin vecinos con bloque -> Inicia un bloque nuevo (+1)
      - Si encuentra un 1 con vecinos con bloque -> Se une al bloque existente
    """
    filas, cols = imagen_binaria.shape
    mapa_bloques = np.zeros_like(imagen_binaria, dtype=int)
    id_bloque = 1
    for i in range(filas):
        for j in range(cols):
            
            # Paso A: Ignorar si es fondo (0)
            if imagen_binaria[i, j] == 0:
                continue
            # Paso B: Consultar vecinos que ya tienen bloque
            vecinos = obtener_vecinos_etiquetados(mapa_bloques, i, j)
            # Paso C: Toma de decisión clara
            if len(vecinos) == 0:
                # Caso 1: No tiene vecinos previos -> GENERAR BLOQUE NUEVO
                id_bloque += 1
                mapa_bloques[i, j] = id_bloque
            else:
                # Caso 2: Tiene vecinos previos -> UNIRSE AL BLOQUE (el menor)
                mapa_bloques[i, j] = np.min(vecinos)
    return mapa_bloques
```

```python

img = np.zeros((8, 8), dtype=int)
img[1:3, 1:3] = 1   # Objeto A
img[5:7, 5:7] = 1   # Objeto B
resultado = segmentar_bloques(img)

print(img)

print(resultado)
```
